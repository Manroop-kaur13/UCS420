### Q2(b). Find the most frequent value and its index/indices

```python
def most_frequent_values_and_indices(array):
    values, counts = np.unique(array, return_counts=True)
    max_count = counts.max()
    most_frequent = values[counts == max_count]

    print("Array:", array)
    print("Most frequent value(s):", most_frequent)
    print("Frequency:", max_count)

    for value in most_frequent:
        indices = np.where(array == value)[0]
        print(f"Indices of {value}:", indices)


# i
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
most_frequent_values_and_indices(x)

print()

# ii
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
most_frequent_values_and_indices(y)
```

## Q3. Access elements of a 2-D array using row and column indices

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("2-D array:\n", arr)

# a) 1st row, 2nd column
# NumPy uses 0-based indexing, so this is arr[0, 1]
print("1st row, 2nd column:", arr[0, 1])

# b) 3rd row, 1st column
# This is arr[2, 0]
print("3rd row, 1st column:", arr[2, 0])
```

## Q4. Create a 1-D array with 25 evenly spaced numbers from 10 to 100

```python
ucs420_Manroop = np.linspace(10, 100, 25)

print("Array:\n", ucs420_Manroop)
print("Dimensions (ndim):", ucs420_Manroop.ndim)
print("Shape:", ucs420_Manroop.shape)
print("Total elements (size):", ucs420_Manroop.size)
print("Data type:", ucs420_Manroop.dtype)
print("Total bytes consumed (nbytes):", ucs420_Manroop.nbytes)
```

### Transpose using reshape()
A 1-D array does not change shape when `.T` is applied. To demonstrate a meaningful transpose, first reshape the 25 elements into a 5×5 matrix and then transpose it.

```python
reshaped_ucs420_Manroop = ucs420_Manroop.reshape(5, 5)
transpose_ucs420_Manroop = reshaped_ucs420_Manroop.T

print("5 x 5 array using reshape():\n", reshaped_ucs420_Manroop)
print("Transpose of the 5 x 5 array using .T:\n", transpose_ucs420_Manroop)

# Can .T be used directly on the original 1-D array?
print("Shape of original 1-D array:", ucs420_Manroop.shape)
print("Shape after applying .T directly:", ucs420_Manroop.T.shape)
print("Therefore, .T does not change a 1-D array's shape.")
```
